```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
!ls
```

```python
df = pd.read_csv("Sample_ Superstore.csv")

print("Dataset Loaded Successfully!")
df.head()
```

```python
print("Shape of Dataset:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nInformation:")
df.info()
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after removing:", df.duplicated().sum())
```

```python
high_sales = df[df["Sales"] > 500]

high_sales.head()
```

```python
df.sort_values(by="Sales", ascending=False).head()
```

```python
print("Sales Statistics")
print("-"*40)
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())

print("\nProfit Statistics")
print("-"*40)
print("Total Profit:", df["Profit"].sum())
print("Average Profit:", df["Profit"].mean())
print("Maximum Profit:", df["Profit"].max())
print("Minimum Profit:", df["Profit"].min())
```

```python
region = df.groupby("Region")[["Sales","Profit"]].sum()

print(region)
```

```python
category = df.groupby("Category")[["Sales","Profit"]].sum()

print(category)
```

```python
category["Sales"].plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
df.groupby("Category")["Sales"].sum().plot(kind="line")

plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
plt.hist(df["Sales"], bins=10)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```
